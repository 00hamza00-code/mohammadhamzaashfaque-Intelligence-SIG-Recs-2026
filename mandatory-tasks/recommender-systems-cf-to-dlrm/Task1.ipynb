```python
import pandas as pd
df = pd.read_csv(
    'ml-100k/u.data', 
    sep='\t', 
    header=None, 
    names=['user_id', 'item_id', 'rating', 'timestamp']
)
```

```python
user_item_matrix = df.pivot(index='user_id', columns='item_id', values='rating').fillna(0)
# print(user_item_matrix.shape) 
# print(user_item_matrix.head())
user_item_matrix.head()
```
